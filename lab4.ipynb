````python
from dotenv import load_dotenv
import os
from openai import OpenAI

load_dotenv()
API_KEY = os.environ["GROQ_API_KEY"]

client = OpenAI(
    api_key=API_KEY,
    base_url="https://api.groq.com/openai/v1"
)

MODEL="llama-3.3-70b-versatile"
print("Client Ready")
print(API_KEY[:10])
````

PART 1.1

````python
def ask_llm(user_prompt,
            system_prompt="You are a helpful assistant.",
            temperature =0.7,
            max_tokens=500):

    response= client.chat.completions.create(
        model = MODEL,
        messages =[
        {"role": "system", "content": system_prompt},
      {"role": "user",   "content": user_prompt},
  ],
        temperature=temperature,
        max_tokens=max_tokens,
    )
    return response.choices[0].message.content
````

They system are instruction given to the model that affect how it behaves for a prompt they don't affect the models weights or parameters just how the model is needed to do it tasks or like the roles they should assume to perform a task. The user gives the task to the model as to what exactly to do. Example System - You are a system designer. Prompt - Design a system for me. A token is how the model reads every character from the prompt so the API charges by token because that what is takes in and also outputs so the prompt is broken down into token and the output is given in token so it chargers per token

````python
import json
import pandas as pd


EXTRACT_PROMPT = """
You are an assistant helping a microfinance loan officer.

Extract information from the loan application.

Return ONLY a valid JSON object with EXACTLY these keys:

{
  "applicant_name": "",
  "amount_ghs": 0,
  "purpose": "",
  "monthly_profit_ghs": null,
  "has_collateral_or_guarantor": false,
  "repayment_months": null
}

Rules:
- Use only information found in the letter.
- If a field is not stated, use null.
- Do not guess.
- Return ONLY valid JSON.

Example:

Letter:
My name is Ama Mensah. I am requesting GHS 5000 to buy a sewing machine.
My monthly profit is GHS 1200.
My mother will guarantee the loan.
I will repay within 10 months.

Output:
{
  "applicant_name": "Ama Mensah",
  "amount_ghs": 5000,
  "purpose": "buy a sewing machine",
  "monthly_profit_ghs": 1200,
  "has_collateral_or_guarantor": true,
  "repayment_months": 10
}
"""

def extract_fields(letter_text):

    user_prompt = f"""
Extract the required information from this loan application.

{letter_text}
"""

    response = ask_llm(
        user_prompt,
        system_prompt=EXTRACT_PROMPT,
        temperature=0
    )

    response = response.replace("```json", "")
    response = response.replace("```", "")
    response = response.strip()

    try:
        return json.loads(response)
    except json.JSONDecodeError:
        print("Warning: Could not parse JSON")
        print(response)
        return None

results = []

for letter_id, letter_text in LETTERS.items():

    data = extract_fields(letter_text)

    if data is not None:
        data["letter_id"] = letter_id
        results.append(data)

df = pd.DataFrame(results)
display(df)
````

The few-shot example should not come from the six letters because the model would already have seen the correct answer, making the extraction unfair and less realistic. The instruction **"use null, do not guess"** prevents the model from inventing information when details such as monthly profit or repayment period are missing, making the extracted data more accurate and reliable. Finally, **temperature = 0** is the best choice for information extraction because it produces consistent and repeatable results, while higher temperatures introduce more randomness, which is useful for creative tasks but not for extracting factual information.

````python


BRIEF_PROMPT = """
You are an AI assistant supporting a microfinance loan officer.

Your job is to assist the loan officer, not make the final decision.

The final lending decision must always be made by a human.

Using BOTH the loan application letter and the extracted JSON, produce a report with the following sections:

1. Strengths
- Use bullet points.
- Only include strengths supported by the letter.

2. Risks / Red Flags
- Use bullet points.
- Do not invent information.

3. Missing Information
- List information the loan officer should request before making a decision.

4. Suggested Next Step
- Suggest an appropriate next step such as:
    - Invite for interview
    - Request supporting documents
    - Request additional financial information
    - Flag for senior review

Never recommend "Approve" or "Reject".
"""

def generate_brief(letter_text, extracted_json):

    user_prompt = f"""
Loan Application Letter:

{letter_text}

Extracted Information:

{json.dumps(extracted_json, indent=2)}

Generate the decision-support brief.
"""

    return ask_llm(
        user_prompt,
        system_prompt=BRIEF_PROMPT,
        temperature=0
    )


briefs = {}

for letter_id, letter_text in LETTERS.items():

    extracted = extract_fields(letter_text)

    if extracted is not None:
        briefs[letter_id] = generate_brief(letter_text, extracted)


print("="*80)
print("L001")
print("="*80)
print(briefs["L001"])

print("\n" + "="*80)
print("L002")
print("="*80)
print(briefs["L002"])

print("\n" + "="*80)
print("L006")
print("="*80)
print(briefs["L006"])
````


The brief for L003 identified strong factors such as a clear business purpose, evidence of repayment ability, and supporting financial information. The brief for L006 highlighted more risks, including missing financial details, weaker evidence of repayment capacity, and additional information that should be requested. This shows the system can distinguish between stronger and weaker applications while remaining grounded in the information provided.
The model may not have all the information needed to make a lending decision, so a human loan officer should review the application before deciding.
Ethical reason: Loan decisions can significantly affect people's lives. A human should remain responsible for the final decision to ensure fairness, accountability, and proper judgment.

````python
# SECTION 4
fields = [
    "applicant_name",
    "amount_ghs",
    "purpose",
    "monthly_profit_ghs",
    "has_collateral_or_guarantor",
    "repayment_months"
]

gold_letters = ["L001", "L003", "L006"]

rows = []

for field in fields:

    row = {"field": field}

    correct = 0

    for letter_id in gold_letters:

        predicted = df.loc[
            df["letter_id"] == letter_id, field
        ].iloc[0]

        expected = GOLD[letter_id][field]

        if field == "applicant_name":
            match = str(predicted).strip().lower() == str(expected).strip().lower()

        else:
            match = predicted == expected

        row[letter_id] = "✓" if match else "✗"

        if match:
            correct += 1

    row["accuracy"] = f"{correct}/3 ({correct/3:.0%})"

    rows.append(row)

accuracy_df = pd.DataFrame(rows)

display(accuracy_df)



````
